```python
import numpy as np
import pandas as pd
import zarr
import os

from scipy.ndimage import gaussian_filter, maximum_filter
from scipy.spatial.distance import cdist

# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

TEST_PATH = os.path.join(DATA_PATH, "test")

TEST_SAMPLES = [
    "44b6_0b24845f",
    "44b6_0113de3b",
    "6bba_05db0fb1",
    "6bba_05b6850b"
]

# Baseline assumptions
MAX_CELLS_PER_FRAME = 10
MAX_LINK_DISTANCE = 12.0

# Physical voxel scales from the dataset
Z_SCALE = 1.625
Y_SCALE = 0.40625
X_SCALE = 0.40625


# --------------------------------------------------
# DETECT CELLS IN ONE FRAME
# --------------------------------------------------

def detect_cells(volume):
    # Smooth the 3D image
    smooth = gaussian_filter(
        volume.astype(np.float32),
        sigma=(1.2, 2.0, 2.0)
    )

    # Find local maxima
    local_max = maximum_filter(
        smooth,
        size=(5, 7, 7)
    )

    mask = (smooth == local_max)

    # Ignore very dim points
    threshold = np.percentile(smooth, 99)

    mask &= smooth > threshold

    coords = np.argwhere(mask)

    if len(coords) == 0:
        return np.empty((0, 3), dtype=int)

    # Rank candidates by intensity
    scores = smooth[coords[:, 0], coords[:, 1], coords[:, 2]]

    order = np.argsort(scores)[::-1]

    coords = coords[order[:MAX_CELLS_PER_FRAME]]

    return coords


# --------------------------------------------------
# PROCESS ONE DATASET
# --------------------------------------------------

def process_dataset(dataset_name):

    path = os.path.join(
        TEST_PATH,
        dataset_name + ".zarr"
    )

    arr = zarr.open(path, mode="r")["0"]

    all_nodes = []
    frame_nodes = {}

    next_node_id = 1

    print(f"\nProcessing {dataset_name}")

    # -----------------------------
    # Detection
    # -----------------------------

    for t in range(arr.shape[0]):

        volume = np.asarray(arr[t])

        coords = detect_cells(volume)

        frame_nodes[t] = []

        for z, y, x in coords:

            node_id = next_node_id
            next_node_id += 1

            all_nodes.append({
                "id": None,
                "dataset": dataset_name,
                "row_type": "node",
                "node_id": node_id,
                "t": t,
                "z": int(z),
                "y": int(y),
                "x": int(x),
                "source_id": -1,
                "target_id": -1
            })

            frame_nodes[t].append(
                (node_id, int(z), int(y), int(x))
            )

    print("Detected nodes:", len(all_nodes))


    # -----------------------------
    # Linking
    # -----------------------------

    edges = []

    for t in range(arr.shape[0] - 1):

        current = frame_nodes[t]
        next_frame = frame_nodes[t + 1]

        if len(current) == 0 or len(next_frame) == 0:
            continue

        current_coords = np.array([
            [
                z * Z_SCALE,
                y * Y_SCALE,
                x * X_SCALE
            ]
            for _, z, y, x in current
        ])

        next_coords = np.array([
            [
                z * Z_SCALE,
                y * Y_SCALE,
                x * X_SCALE
            ]
            for _, z, y, x in next_frame
        ])

        distances = cdist(current_coords, next_coords)

        for i, (node_id, z, y, x) in enumerate(current):

            j = np.argmin(distances[i])

            if distances[i, j] <= MAX_LINK_DISTANCE:

                target_id = next_frame[j][0]

                edges.append({
                    "id": None,
                    "dataset": dataset_name,
                    "row_type": "edge",
                    "node_id": -1,
                    "t": -1,
                    "z": -1,
                    "y": -1,
                    "x": -1,
                    "source_id": node_id,
                    "target_id": target_id
                })

    print("Detected edges:", len(edges))

    return all_nodes + edges


# --------------------------------------------------
# RUN ALL FOUR TEST DATASETS
# --------------------------------------------------

rows = []

for dataset_name in TEST_SAMPLES:
    rows.extend(process_dataset(dataset_name))


# --------------------------------------------------
# CREATE SUBMISSION
# --------------------------------------------------

submission = pd.DataFrame(rows)

submission["id"] = np.arange(len(submission))

submission = submission[
    [
        "id",
        "dataset",
        "row_type",
        "node_id",
        "t",
        "z",
        "y",
        "x",
        "source_id",
        "target_id"
    ]
]

print("\n==============================")
print("Submission shape:", submission.shape)
print("==============================")

display(submission.head(20))

submission_path = "/kaggle/working/submission.csv"

submission.to_csv(
    submission_path,
    index=False
)

print("\nSaved:", submission_path)
```

```python
import numpy as np
import pandas as pd
import zarr
import os

from scipy.ndimage import gaussian_filter, maximum_filter
from scipy.spatial.distance import cdist


# --------------------------------------------------
# SETTINGS
# --------------------------------------------------

TEST_PATH = os.path.join(DATA_PATH, "test")

TEST_SAMPLES = [
    "44b6_0b24845f",
    "44b6_0113de3b",
    "6bba_05db0fb1",
    "6bba_05b6850b"
]

MAX_CELLS_PER_FRAME = 15
MIN_CELL_SCORE_PERCENTILE = 99.3
MIN_BORDER = 3
MAX_LINK_DISTANCE = 12.0

Z_SCALE = 1.625
Y_SCALE = 0.40625
X_SCALE = 0.40625


# --------------------------------------------------
# DETECTION
# --------------------------------------------------

def detect_cells(volume):

    smooth = gaussian_filter(
        volume.astype(np.float32),
        sigma=(1.2, 2.0, 2.0)
    )

    local_max = maximum_filter(
        smooth,
        size=(5, 7, 7)
    )

    mask = (smooth == local_max)

    threshold = np.percentile(
        smooth,
        MIN_CELL_SCORE_PERCENTILE
    )

    mask &= smooth > threshold

    coords = np.argwhere(mask)

    if len(coords) == 0:
        return np.empty((0, 3), dtype=int)

    # Remove detections too close to volume boundaries
    z, y, x = coords.T

    valid = (
        (z >= MIN_BORDER) &
        (z < volume.shape[0] - MIN_BORDER) &
        (y >= MIN_BORDER) &
        (y < volume.shape[1] - MIN_BORDER) &
        (x >= MIN_BORDER) &
        (x < volume.shape[2] - MIN_BORDER)
    )

    coords = coords[valid]

    if len(coords) == 0:
        return np.empty((0, 3), dtype=int)

    scores = smooth[
        coords[:, 0],
        coords[:, 1],
        coords[:, 2]
    ]

    order = np.argsort(scores)[::-1]

    coords = coords[
        order[:MAX_CELLS_PER_FRAME]
    ]

    return coords


# --------------------------------------------------
# PROCESS DATASET
# --------------------------------------------------

def process_dataset(dataset_name):

    path = os.path.join(
        TEST_PATH,
        dataset_name + ".zarr"
    )

    arr = zarr.open(path, mode="r")["0"]

    all_nodes = []
    frame_nodes = {}

    next_node_id = 1

    print(f"\nProcessing {dataset_name}")

    # -----------------------------
    # Detect nodes
    # -----------------------------

    for t in range(arr.shape[0]):

        coords = detect_cells(arr[t])

        frame_nodes[t] = []

        for z, y, x in coords:

            node_id = next_node_id
            next_node_id += 1

            all_nodes.append({
                "id": None,
                "dataset": dataset_name,
                "row_type": "node",
                "node_id": node_id,
                "t": t,
                "z": int(z),
                "y": int(y),
                "x": int(x),
                "source_id": -1,
                "target_id": -1
            })

            frame_nodes[t].append(
                (node_id, int(z), int(y), int(x))
            )

    print("Detected nodes:", len(all_nodes))

    # -----------------------------
    # One-to-one linking
    # -----------------------------

    edges = []

    for t in range(arr.shape[0] - 1):

        current = frame_nodes[t]
        next_frame = frame_nodes[t + 1]

        if not current or not next_frame:
            continue

        current_coords = np.array([
            [z * Z_SCALE, y * Y_SCALE, x * X_SCALE]
            for _, z, y, x in current
        ])

        next_coords = np.array([
            [z * Z_SCALE, y * Y_SCALE, x * X_SCALE]
            for _, z, y, x in next_frame
        ])

        distances = cdist(
            current_coords,
            next_coords
        )

        # Create all possible candidate links
        candidates = []

        for i in range(len(current)):
            for j in range(len(next_frame)):

                distance = distances[i, j]

                if distance <= MAX_LINK_DISTANCE:
                    candidates.append(
                        (distance, i, j)
                    )

        # Closest links first
        candidates.sort()

        used_current = set()
        used_next = set()

        for distance, i, j in candidates:

            if i in used_current:
                continue

            if j in used_next:
                continue

            source_id = current[i][0]
            target_id = next_frame[j][0]

            edges.append({
                "id": None,
                "dataset": dataset_name,
                "row_type": "edge",
                "node_id": -1,
                "t": -1,
                "z": -1,
                "y": -1,
                "x": -1,
                "source_id": source_id,
                "target_id": target_id
            })

            used_current.add(i)
            used_next.add(j)

    print("Detected edges:", len(edges))

    return all_nodes + edges


# --------------------------------------------------
# RUN
# --------------------------------------------------

rows = []

for dataset_name in TEST_SAMPLES:
    rows.extend(
        process_dataset(dataset_name)
    )


# --------------------------------------------------
# CREATE SUBMISSION
# --------------------------------------------------

submission = pd.DataFrame(rows)

submission["id"] = np.arange(
    len(submission)
)

submission = submission[
    [
        "id",
        "dataset",
        "row_type",
        "node_id",
        "t",
        "z",
        "y",
        "x",
        "source_id",
        "target_id"
    ]
]

submission_path = "/kaggle/working/submission.csv"

submission.to_csv(
    submission_path,
    index=False
)

print("\n==============================")
print("FINAL BASELINE")
print("==============================")
print("Submission shape:", submission.shape)
print("Nodes:", (submission["row_type"] == "node").sum())
print("Edges:", (submission["row_type"] == "edge").sum())
print("Saved:", submission_path)

display(submission.head(10))
```

```python
from scipy.spatial.distance import cdist

predicted_motion = []

# Store node positions by time
nodes_by_time = {
    t_value: group.copy()
    for t_value, group in gt.groupby("t")
}

times = sorted(nodes_by_time.keys())

# First transition: no previous position available,
# so use nearest-neighbor.
current = nodes_by_time[times[0]]
next_frame = nodes_by_time[times[1]]

current_xyz = current[["z", "y", "x"]].to_numpy() * np.array([
    Z_SCALE, Y_SCALE, X_SCALE
])

next_xyz = next_frame[["z", "y", "x"]].to_numpy() * np.array([
    Z_SCALE, Y_SCALE, X_SCALE
])

distances = cdist(current_xyz, next_xyz)
nearest_indices = np.argmin(distances, axis=1)

for i, j in enumerate(nearest_indices):
    predicted_motion.append((
        current.iloc[i]["node_id"],
        next_frame.iloc[j]["node_id"]
    ))


# Remaining transitions: use motion prediction
for k in range(1, len(times) - 1):

    previous = nodes_by_time[times[k - 1]]
    current = nodes_by_time[times[k]]
    next_frame = nodes_by_time[times[k + 1]]

    # Position lookup
    previous_pos = previous.set_index("node_id")[["z", "y", "x"]]
    current_pos = current.set_index("node_id")[["z", "y", "x"]]

    next_xyz = next_frame[["z", "y", "x"]].to_numpy() * np.array([
        Z_SCALE, Y_SCALE, X_SCALE
    ])

    for _, row in current.iterrows():

        current_id = row["node_id"]

        # Find this node's previous position.
        # If it has no previous position, skip motion prediction.
        previous_candidates = previous[
            previous["t"] == times[k - 1]
        ]

        # Find the nearest previous node
        current_xyz_single = row[["z", "y", "x"]].to_numpy(dtype=float)
        previous_xyz = previous_candidates[
            ["z", "y", "x"]
        ].to_numpy(dtype=float)

        previous_xyz_physical = previous_xyz * np.array([
            Z_SCALE, Y_SCALE, X_SCALE
        ])

        current_xyz_physical = current_xyz_single * np.array([
            Z_SCALE, Y_SCALE, X_SCALE
        ])

        prev_distances = np.linalg.norm(
            previous_xyz_physical - current_xyz_physical,
            axis=1
        )

        previous_idx = np.argmin(prev_distances)
        previous_row = previous_candidates.iloc[previous_idx]

        # Estimate velocity
        velocity = (
            current_xyz_physical
            - previous_row[["z", "y", "x"]].to_numpy(dtype=float)
              * np.array([Z_SCALE, Y_SCALE, X_SCALE])
        )

        # Predict next position
        predicted_position = current_xyz_physical + velocity

        # Find closest next-frame cell to prediction
        distances_to_prediction = np.linalg.norm(
            next_xyz - predicted_position,
            axis=1
        )

        next_idx = np.argmin(distances_to_prediction)

        predicted_motion.append((
            current_id,
            next_frame.iloc[next_idx]["node_id"]
        ))

print("Predicted motion-aware edges:", len(predicted_motion))
```

```python
# Adaptive Motion-Aware Tracking

from scipy.spatial.distance import cdist
import numpy as np

predicted_adaptive = []

times = sorted(nodes_by_time.keys())

for k in range(len(times) - 1):

    current = nodes_by_time[times[k]]
    next_frame = nodes_by_time[times[k + 1]]

    current_xyz = current[["z", "y", "x"]].to_numpy() * np.array([
        Z_SCALE, Y_SCALE, X_SCALE
    ])

    next_xyz = next_frame[["z", "y", "x"]].to_numpy() * np.array([
        Z_SCALE, Y_SCALE, X_SCALE
    ])

    # First frame: no motion history
    if k == 0:

        distances = cdist(current_xyz, next_xyz)
        nearest_indices = np.argmin(distances, axis=1)

        for i, j in enumerate(nearest_indices):
            predicted_adaptive.append((
                current.iloc[i]["node_id"],
                next_frame.iloc[j]["node_id"]
            ))

        continue

    previous = nodes_by_time[times[k - 1]]

    previous_xyz = previous[["z", "y", "x"]].to_numpy() * np.array([
        Z_SCALE, Y_SCALE, X_SCALE
    ])

    # Match each current cell to its nearest previous cell
    previous_distances = cdist(current_xyz, previous_xyz)
    previous_indices = np.argmin(previous_distances, axis=1)

    for i, prev_idx in enumerate(previous_indices):

        current_position = current_xyz[i]
        previous_position = previous_xyz[prev_idx]

        # Estimate velocity
        velocity = current_position - previous_position

        # Measure how reliable that previous movement was
        movement_size = np.linalg.norm(velocity)

        # Adaptive rule:
        # If movement is reasonable, use motion prediction.
        # Otherwise use current position.
        if movement_size <= 4.2:

            predicted_position = current_position + velocity

        else:

            predicted_position = current_position

        # Find closest next-frame cell
        distances = np.linalg.norm(
            next_xyz - predicted_position,
            axis=1
        )

        next_idx = np.argmin(distances)

        predicted_adaptive.append((
            current.iloc[i]["node_id"],
            next_frame.iloc[next_idx]["node_id"]
        ))

print("Predicted adaptive edges:", len(predicted_adaptive))
```

```python
adaptive_edges = set(
    map(tuple, predicted_adaptive)
)

true_positives = len(
    adaptive_edges & ground_truth_edges
)

precision = true_positives / len(adaptive_edges)
recall = true_positives / len(ground_truth_edges)
f1 = 2 * precision * recall / (precision + recall)

print("Adaptive Motion-Aware Results")
print("-----------------------------")
print("True positives:", true_positives)
print("Predicted edges:", len(adaptive_edges))
print("Ground-truth edges:", len(ground_truth_edges))
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1:", round(f1, 4))
```

```python
MAX_CELLS_PER_FRAME = 15
THRESHOLD_PERCENTILE = 99.3
MIN_BORDER = 3

detected_by_time = {}

for t_value in range(volume.shape[0]):

    frame = np.asarray(volume[t_value])
    # Smooth the 3D image
    smooth = gaussian_filter(
        frame.astype(np.float32),
        sigma=(1.2, 2.0, 2.0)
    )

    # Find local maxima
    local_max = maximum_filter(
        smooth,
        size=(5, 7, 7)
    )

    threshold = np.percentile(
        smooth,
        THRESHOLD_PERCENTILE
    )

    mask = (
        (smooth == local_max) &
        (smooth >= threshold)
    )

    coords = np.argwhere(mask)

    # Remove detections too close to borders
    valid = (
        (coords[:, 0] >= MIN_BORDER) &
        (coords[:, 0] < frame.shape[0] - MIN_BORDER) &
        (coords[:, 1] >= MIN_BORDER) &
        (coords[:, 1] < frame.shape[1] - MIN_BORDER) &
        (coords[:, 2] >= MIN_BORDER) &
        (coords[:, 2] < frame.shape[2] - MIN_BORDER)
    )

    coords = coords[valid]

    # Keep strongest detections
    if len(coords) > MAX_CELLS_PER_FRAME:

        intensities = smooth[
            coords[:, 0],
            coords[:, 1],
            coords[:, 2]
        ]

        strongest = np.argsort(intensities)[
            -MAX_CELLS_PER_FRAME:
        ]

        coords = coords[strongest]

    detected_by_time[t_value] = coords

print("Detection complete.")

print("\nExample:")
for t_value in range(5):
    print(
        f"t={t_value}: "
        f"{len(detected_by_time[t_value])} detections"
    )
```

```python
from scipy.spatial.distance import cdist

MATCH_DISTANCE = 3.0

total_tp = 0
total_fp = 0
total_fn = 0

for t_value in range(volume.shape[0]):

    # Ground-truth cells at this frame
    gt_frame = gt[gt["t"] == t_value]

    gt_xyz = gt_frame[["z", "y", "x"]].to_numpy() * np.array([
        Z_SCALE, Y_SCALE, X_SCALE
    ])

    # Our detected cells
    detections = detected_all[t_value]

    detected_xyz = detections * np.array([
        Z_SCALE, Y_SCALE, X_SCALE
    ])

    if len(detected_xyz) == 0:
        total_fn += len(gt_xyz)
        continue

    if len(gt_xyz) == 0:
        total_fp += len(detected_xyz)
        continue

    # Distance between every detection and GT cell
    distances = cdist(detected_xyz, gt_xyz)

    # Greedy one-to-one matching
    candidate_pairs = []

    for i in range(len(detected_xyz)):
        for j in range(len(gt_xyz)):
            if distances[i, j] <= MATCH_DISTANCE:
                candidate_pairs.append(
                    (distances[i, j], i, j)
                )

    # Match closest pairs first
    candidate_pairs.sort()

    matched_detections = set()
    matched_gt = set()

    for distance, i, j in candidate_pairs:

        if i in matched_detections:
            continue

        if j in matched_gt:
            continue

        matched_detections.add(i)
        matched_gt.add(j)

    tp = len(matched_detections)
    fp = len(detected_xyz) - tp
    fn = len(gt_xyz) - tp

    total_tp += tp
    total_fp += fp
    total_fn += fn

precision = total_tp / (total_tp + total_fp)
recall = total_tp / (total_tp + total_fn)
f1 = 2 * precision * recall / (precision + recall)

print("Detection Evaluation")
print("--------------------")
print("True positives:", total_tp)
print("False positives:", total_fp)
print("False negatives:", total_fn)
print("Precision:", round(precision, 4))
print("Recall:", round(recall, 4))
print("F1:", round(f1, 4))
```

```python
# ============================================
# 3D LoG CELL DETECTOR
# ============================================

def detect_cells_log(volume, sigma=2.0, threshold_percentile=99.0,
                     min_distance=3):

    # Convert Zarr array to NumPy, then to float
    volume = np.asarray(volume).astype(np.float32)

    # Normalize volume
    vmin = volume.min()
    vmax = volume.max()

    if vmax > vmin:
        volume_norm = (volume - vmin) / (vmax - vmin)
    else:
        volume_norm = volume

    # 3D Laplacian of Gaussian
    log_response = -gaussian_laplace(
        volume_norm,
        sigma=sigma
    )

    # Threshold LoG response
    threshold = np.percentile(
        log_response,
        threshold_percentile
    )

    candidates = log_response > threshold

    # Find local maxima
    neighborhood = 2 * min_distance + 1

    local_max = (
        log_response ==
        maximum_filter(
            log_response,
            size=neighborhood
        )
    )

    detections = candidates & local_max

    # Get coordinates
    coords = np.argwhere(detections)

    # Sort strongest responses first
    if len(coords) > 0:
        scores = log_response[
            coords[:, 0],
            coords[:, 1],
            coords[:, 2]
        ]

        order = np.argsort(scores)[::-1]
        coords = coords[order]

    # Non-maximum suppression using minimum distance
    selected = []

    for point in coords:
        if not selected:
            selected.append(point)
            continue

        distances = np.linalg.norm(
            np.asarray(selected) - point,
            axis=1
        )

        if np.all(distances >= min_distance):
            selected.append(point)

    return np.asarray(selected)
```

```python
import os
import numpy as np
import zarr

SCALE = np.array([1.625, 0.40625, 0.40625])  # (z, y, x) µm/voxel

def load_geff_stats(geff_path):
    g = zarr.open(geff_path, mode="r")

    node_ids = np.asarray(g["nodes"]["ids"])
    edges = np.asarray(g["edges"]["ids"])

    props = zarr.open(f"{geff_path}/nodes/props", mode="r")

    x = np.asarray(props["x"]["values"])
    y = np.asarray(props["y"]["values"])
    z = np.asarray(props["z"]["values"])
    t = np.asarray(props["t"]["values"])

    coords = np.column_stack([z, y, x]).astype(float)

    # Map GEFF node ID -> row index
    id_to_idx = {int(node_id): i for i, node_id in enumerate(node_ids)}

    # Convert edge node IDs into row indices
    E = np.array(
        [[id_to_idx[int(a)], id_to_idx[int(b)]] for a, b in edges],
        dtype=int
    )

    # -------------------------
    # Division statistics
    # -------------------------
    if len(E) > 0:
        out_degree = np.bincount(E[:, 0], minlength=len(node_ids))
        divisions = int((out_degree >= 2).sum())
    else:
        divisions = 0

    # -------------------------
    # Link displacement
    # -------------------------
    if len(E) > 0:
        src = E[:, 0]
        dst = E[:, 1]

        # Only temporal links
        temporal = t[dst] > t[src]

        displacements = np.linalg.norm(
            (coords[dst[temporal]] - coords[src[temporal]]) * SCALE,
            axis=1
        )
    else:
        displacements = np.array([])

    return {
        "nodes": len(node_ids),
        "edges": len(E),
        "divisions": divisions,
        "displacements": displacements,
    }


# ============================================
# Run over all 199 training volumes
# ============================================

train_dir = f"{BASE}/train"

geff_files = sorted(
    f for f in os.listdir(train_dir)
    if f.endswith(".geff")
)

all_displacements = []

total_nodes = 0
total_edges = 0
total_divisions = 0
volumes_with_divisions = 0

prefix_stats = {
    "44b6": {
        "volumes": 0,
        "nodes": 0,
        "edges": 0,
        "divisions": 0,
        "volumes_with_divisions": 0,
        "displacements": []
    },
    "6bba": {
        "volumes": 0,
        "nodes": 0,
        "edges": 0,
        "divisions": 0,
        "volumes_with_divisions": 0,
        "displacements": []
    }
}


for filename in geff_files:

    path = os.path.join(train_dir, filename)

    stats = load_geff_stats(path)

    total_nodes += stats["nodes"]
    total_edges += stats["edges"]
    total_divisions += stats["divisions"]

    if stats["divisions"] > 0:
        volumes_with_divisions += 1

    all_displacements.extend(stats["displacements"])

    # Prefix
    if filename.startswith("44b6"):
        prefix = "44b6"
    elif filename.startswith("6bba"):
        prefix = "6bba"
    else:
        continue

    prefix_stats[prefix]["volumes"] += 1
    prefix_stats[prefix]["nodes"] += stats["nodes"]
    prefix_stats[prefix]["edges"] += stats["edges"]
    prefix_stats[prefix]["divisions"] += stats["divisions"]

    if stats["divisions"] > 0:
        prefix_stats[prefix]["volumes_with_divisions"] += 1

    prefix_stats[prefix]["displacements"].extend(
        stats["displacements"]
    )


# ============================================
# Final summary
# ============================================

all_displacements = np.asarray(all_displacements)

print("=" * 60)
print("GLOBAL GEFF GRAPH STATISTICS")
print("=" * 60)

print(f"Training volumes:        {len(geff_files)}")
print(f"Total annotated nodes:   {total_nodes}")
print(f"Total edges:             {total_edges}")
print(f"Total division nodes:    {total_divisions}")
print(f"Volumes with divisions:  {volumes_with_divisions}")

if len(all_displacements) > 0:
    print("\nLINK DISPLACEMENT (µm)")
    print(f"p50:  {np.percentile(all_displacements, 50):.2f}")
    print(f"p90:  {np.percentile(all_displacements, 90):.2f}")
    print(f"p99:  {np.percentile(all_displacements, 99):.2f}")
    print(f"max:  {all_displacements.max():.2f}")


# ============================================
# Prefix summaries
# ============================================

print("\n" + "=" * 60)
print("PREFIX STATISTICS")
print("=" * 60)

for prefix in ["44b6", "6bba"]:

    s = prefix_stats[prefix]

    print(f"\n{prefix}")
    print("-" * 30)
    print(f"Volumes:               {s['volumes']}")
    print(f"Nodes:                 {s['nodes']}")
    print(f"Edges:                 {s['edges']}")
    print(f"Division nodes:        {s['divisions']}")
    print(f"Volumes with divisions:{s['volumes_with_divisions']}")

    d = np.asarray(s["displacements"])

    if len(d) > 0:
        print(f"Link p50 (µm):         {np.percentile(d, 50):.2f}")
        print(f"Link p90 (µm):         {np.percentile(d, 90):.2f}")
        print(f"Link p99 (µm):         {np.percentile(d, 99):.2f}")
        print(f"Link max (µm):         {d.max():.2f}")
```

```python
# ============================================================
# CELL 8A — END-TO-END DETECTOR + TRACKER FUNCTIONS
# ============================================================

import numpy as np
import zarr

from scipy.ndimage import gaussian_filter, maximum_filter
from scipy.spatial.distance import cdist
from scipy.optimize import linear_sum_assignment


# ------------------------------------------------------------
# Physical scale
# ------------------------------------------------------------

SCALE = np.array([
    1.625,    # Z: µm / voxel
    0.40625,  # Y: µm / voxel
    0.40625   # X: µm / voxel
])


# ------------------------------------------------------------
# Downsampling
# ------------------------------------------------------------

DS = 2

SCALE_DS = SCALE * np.array([1, DS, DS])


# ------------------------------------------------------------
# Physical-scale LoG
# ------------------------------------------------------------

def log_phys(volume, sigma_um, scale):

    sig = [sigma_um / s for s in scale]

    out = np.zeros_like(volume, dtype=np.float32)

    for ax in range(3):

        order = [0, 0, 0]
        order[ax] = 2

        out += (
            gaussian_filter(
                volume,
                sig,
                order=order,
                mode="nearest"
            )
            / scale[ax]**2
        )

    return -(sigma_um**2) * out


# ------------------------------------------------------------
# Peak detection
# ------------------------------------------------------------

def peaks_phys(response, scale, min_dist_um=4.0):

    size = tuple(
        int(2 * np.ceil(min_dist_um / s) + 1)
        for s in scale
    )

    peaks = (
        (response == maximum_filter(
            response,
            size=size,
            mode="nearest"
        ))
        &
        (response > 0)
    )

    coords = np.argwhere(peaks)
    scores = response[peaks]

    order = np.argsort(-scores)

    return coords[order], scores[order]


# ------------------------------------------------------------
# Detect cells in one 3D frame
# ------------------------------------------------------------

def detect_frame(volume,
                 sigma_um=3.0,
                 min_dist_um=4.0):

    volume = np.asarray(volume).astype(np.float32)
    # Normalize intensity
    vmin = volume.min()
    vmax = volume.max()

    if vmax > vmin:
        volume = (volume - vmin) / (vmax - vmin)

    response = log_phys(
        volume,
        sigma_um=sigma_um,
        scale=SCALE_DS
    )

    coords, scores = peaks_phys(
        response,
        scale=SCALE_DS,
        min_dist_um=min_dist_um
    )

    # Convert voxel coordinates → physical µm coordinates
    coords_um = coords * SCALE_DS

    return coords, coords_um, scores


# ------------------------------------------------------------
# Frame-to-frame tracking
# ------------------------------------------------------------

def link_frames(detections_um, gate_um=8.0):

    offsets = np.cumsum(
        [0] + [len(d) for d in detections_um]
    )

    edges = []

    for t in range(len(detections_um) - 1):

        a = detections_um[t]
        b = detections_um[t + 1]

        if len(a) == 0 or len(b) == 0:
            continue

        distances = cdist(a, b)

        cost = np.where(
            distances <= gate_um,
            distances,
            1e6
        )

        rows, cols = linear_sum_assignment(cost)

        for i, j in zip(rows, cols):

            if cost[i, j] < 1e6:

                edges.append(
                    (
                        offsets[t] + i,
                        offsets[t + 1] + j
                    )
                )

    return np.asarray(
        edges,
        dtype=int
    ).reshape(-1, 2)


print("Cell 8A loaded successfully.")
print("Detector: physical LoG, sigma = 3 µm")
print("Minimum detection distance = 4 µm")
print("Tracking gate = 8 µm")
print("Downsampling = 2× XY")
```

```python
# Cell 9B — Run baseline on held-out volumes

def load_gt_for_volume(geff_path):
    props = zarr.open(f"{geff_path}/nodes/props", mode="r")

    t = np.asarray(props["t"]["values"])
    z = np.asarray(props["z"]["values"])
    y = np.asarray(props["y"]["values"])
    x = np.asarray(props["x"]["values"])

    xyz = np.column_stack([z, y, x])

    return t, xyz


def match_nodes(pred_um, gt_vox, radius_um=7.0):
    """
    Match predicted physical coordinates against GT voxel coordinates.
    Returns:
        matched_gt_indices
        mean localization error
    """

    if len(pred_um) == 0 or len(gt_vox) == 0:
        return [], []

    gt_um = gt_vox * SCALE

    d = cdist(pred_um, gt_um)

    cost = np.where(d <= radius_um, d, 1e6)

    rows, cols = linear_sum_assignment(cost)

    matches = []
    errors = []

    for r, c in zip(rows, cols):
        if cost[r, c] < 1e6:
            matches.append(c)
            errors.append(cost[r, c])

    return matches, errors


def evaluate_volume(zarr_path, frame_ids=(0, 33, 66, 99)):

    name = os.path.basename(zarr_path).replace(".zarr", "")
    geff_path = f"{train_dir}/{name}.geff"

    if not os.path.exists(geff_path):
        print(f"Skipping {name}: GEFF not found")
        return None

    z = zarr.open(zarr_path, mode="r")
    volume = z["0"]

    gt_t, gt_xyz = load_gt_for_volume(geff_path)

    rows = []

    for t in frame_ids:

        frame = np.asarray(volume[t])

        # Downsample XY exactly as our baseline does
        frame_ds = frame[:, ::DS, ::DS]

        coords_vox, coords_um, scores = detect_frame(
            frame_ds,
            sigma_um=SIGMA_UM,
            min_dist_um=MIN_DIST_UM
        )

        gt_mask = gt_t == t
        gt_frame = gt_xyz[gt_mask]

        matches, errors = match_nodes(
            coords_um,
            gt_frame,
            radius_um=7.0
        )

        recall = (
            len(matches) / len(gt_frame)
            if len(gt_frame) > 0
            else np.nan
        )

        rows.append({
            "volume": name,
            "prefix": "44b6" if "44b6_" in name else "6bba",
            "frame": t,
            "pred_nodes": len(coords_um),
            "gt_nodes": len(gt_frame),
            "node_recall": recall,
            "mean_error_um": (
                np.mean(errors) if errors else np.nan
            )
        })

    return rows


results = []

for path in selected_paths:

    print("\nProcessing:", os.path.basename(path))

    r = evaluate_volume(path)

    if r is not None:
        results.extend(r)

results_df = pd.DataFrame(results)

results_df
```

```python
# ============================================================
# BIOHUB — FINAL SUBMISSION PIPELINE
# ============================================================

import os
import numpy as np
import pandas as pd
import zarr

from scipy.ndimage import gaussian_filter, maximum_filter
from scipy.spatial.distance import cdist


# ============================================================
# 1. PATHS
# ============================================================

DATA_PATH = "/kaggle/input/competitions/biohub-cell-tracking-during-development"

TEST_PATH = os.path.join(DATA_PATH, "test")

TEST_SAMPLES = [
    "44b6_0b24845f",
    "44b6_0113de3b",
    "6bba_05db0fb1",
    "6bba_05b6850b"
]

print("DATA_PATH:", DATA_PATH)
print("TEST_PATH:", TEST_PATH)
print()


# ============================================================
# 2. SETTINGS
# ============================================================

MAX_CELLS_PER_FRAME = 15
MIN_CELL_SCORE_PERCENTILE = 99.3
MIN_BORDER = 3
MAX_LINK_DISTANCE = 12.0

# Physical voxel scaling
Z_SCALE = 1.625
Y_SCALE = 0.40625
X_SCALE = 0.40625


# ============================================================
# 3. CELL DETECTOR
# ============================================================

def detect_cells(volume):

    smooth = gaussian_filter(
        volume.astype(np.float32),
        sigma=(1.2, 2.0, 2.0)
    )

    local_max = maximum_filter(
        smooth,
        size=(5, 7, 7)
    )

    mask = (smooth == local_max)

    threshold = np.percentile(
        smooth,
        MIN_CELL_SCORE_PERCENTILE
    )

    mask &= smooth > threshold

    coords = np.argwhere(mask)

    if len(coords) == 0:
        return np.empty((0, 3), dtype=int)

    # Remove detections too close to volume boundaries
    z, y, x = coords.T

    valid = (
        (z >= MIN_BORDER) &
        (z < volume.shape[0] - MIN_BORDER) &
        (y >= MIN_BORDER) &
        (y < volume.shape[1] - MIN_BORDER) &
        (x >= MIN_BORDER) &
        (x < volume.shape[2] - MIN_BORDER)
    )

    coords = coords[valid]

    if len(coords) == 0:
        return np.empty((0, 3), dtype=int)

    # Score detections
    scores = smooth[
        coords[:, 0],
        coords[:, 1],
        coords[:, 2]
    ]

    # Keep strongest detections
    order = np.argsort(scores)[::-1]

    coords = coords[
        order[:MAX_CELLS_PER_FRAME]
    ]

    return coords


# ============================================================
# 4. PROCESS ONE DATASET
# ============================================================

def process_dataset(dataset_name):

    path = os.path.join(
        TEST_PATH,
        dataset_name + ".zarr"
    )

    print("\n========================================")
    print("Processing:", dataset_name)
    print("Path:", path)
    print("========================================")

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Dataset not found: {path}"
        )

    arr = zarr.open(path, mode="r")["0"]

    print("Volume shape:", arr.shape)

    all_nodes = []
    frame_nodes = {}

    next_node_id = 1

    # --------------------------------------------------------
    # Detect nodes
    # --------------------------------------------------------

    for t in range(arr.shape[0]):

        coords = detect_cells(arr[t])

        frame_nodes[t] = []

        for z, y, x in coords:

            node_id = next_node_id
            next_node_id += 1

            all_nodes.append({
                "id": None,
                "dataset": dataset_name,
                "row_type": "node",
                "node_id": node_id,
                "t": t,
                "z": int(z),
                "y": int(y),
                "x": int(x),
                "source_id": -1,
                "target_id": -1
            })

            frame_nodes[t].append(
                (
                    node_id,
                    int(z),
                    int(y),
                    int(x)
                )
            )

        # Progress
        if (t + 1) % 10 == 0 or t == arr.shape[0] - 1:
            print(
                f"Frame {t + 1}/{arr.shape[0]} | "
                f"nodes so far: {len(all_nodes)}"
            )

    print("\nDetected nodes:", len(all_nodes))

    # --------------------------------------------------------
    # One-to-one linking
    # --------------------------------------------------------

    edges = []

    for t in range(arr.shape[0] - 1):

        current = frame_nodes[t]
        next_frame = frame_nodes[t + 1]

        if not current or not next_frame:
            continue

        current_coords = np.array([
            [
                z * Z_SCALE,
                y * Y_SCALE,
                x * X_SCALE
            ]
            for _, z, y, x in current
        ])

        next_coords = np.array([
            [
                z * Z_SCALE,
                y * Y_SCALE,
                x * X_SCALE
            ]
            for _, z, y, x in next_frame
        ])

        distances = cdist(
            current_coords,
            next_coords
        )

        # Candidate links
        candidates = []

        for i in range(len(current)):
            for j in range(len(next_frame)):

                distance = distances[i, j]

                if distance <= MAX_LINK_DISTANCE:
                    candidates.append(
                        (distance, i, j)
                    )

        # Closest links first
        candidates.sort()

        used_current = set()
        used_next = set()

        for distance, i, j in candidates:

            if i in used_current:
                continue

            if j in used_next:
                continue

            source_id = current[i][0]
            target_id = next_frame[j][0]

            edges.append({
                "id": None,
                "dataset": dataset_name,
                "row_type": "edge",
                "node_id": -1,
                "t": -1,
                "z": -1,
                "y": -1,
                "x": -1,
                "source_id": source_id,
                "target_id": target_id
            })

            used_current.add(i)
            used_next.add(j)

    print("Detected edges:", len(edges))

    return all_nodes + edges


# ============================================================
# 5. RUN ALL FOUR TEST DATASETS
# ============================================================

rows = []

for dataset_name in TEST_SAMPLES:

    dataset_rows = process_dataset(
        dataset_name
    )

    rows.extend(dataset_rows)

    print(
        f"\nFinished {dataset_name}: "
        f"{len(dataset_rows)} rows"
    )


# ============================================================
# 6. CREATE SUBMISSION
# ============================================================

submission = pd.DataFrame(rows)

submission["id"] = np.arange(
    len(submission)
)

submission = submission[
    [
        "id",
        "dataset",
        "row_type",
        "node_id",
        "t",
        "z",
        "y",
        "x",
        "source_id",
        "target_id"
    ]
]


# ============================================================
# 7. SAVE SUBMISSION
# ============================================================

submission_path = "/kaggle/working/submission.csv"

submission.to_csv(
    submission_path,
    index=False
)


# ============================================================
# 8. BASIC VALIDATION
# ============================================================

print("\n")
print("========================================")
print("FINAL SUBMISSION")
print("========================================")

print("Submission shape:", submission.shape)

print(
    "Nodes:",
    (submission["row_type"] == "node").sum()
)

print(
    "Edges:",
    (submission["row_type"] == "edge").sum()
)

print("Saved:", submission_path)


print("\nColumns:")
print(list(submission.columns))


print("\nMissing values:")
print(submission.isna().sum())


print("\nRow IDs sequential:")
print(
    submission["id"].to_numpy().tolist()
    == list(range(len(submission)))
)


print("\nDataset counts:")

print(
    submission
    .groupby(
        ["dataset", "row_type"]
    )
    .size()
    .unstack(fill_value=0)
)


# ============================================================
# 9. NODE VALIDATION
# ============================================================

node_rows = submission[
    submission["row_type"] == "node"
]

print("\nNode coordinate ranges:")

print(
    "z:",
    node_rows["z"].min(),
    "to",
    node_rows["z"].max()
)

print(
    "y:",
    node_rows["y"].min(),
    "to",
    node_rows["y"].max()
)

print(
    "x:",
    node_rows["x"].min(),
    "to",
    node_rows["x"].max()
)


# ============================================================
# 10. DUPLICATE NODE CHECK
# ============================================================

duplicates = (
    node_rows
    .groupby(
        ["dataset", "node_id"]
    )
    .size()
)

print("\nDuplicate node IDs:")
print(
    (duplicates > 1).sum()
)


# ============================================================
# 11. EDGE VALIDATION
# ============================================================

print("\nEdge validation:")

for dataset in submission["dataset"].unique():

    nodes = set(
        node_rows[
            node_rows["dataset"] == dataset
        ]["node_id"]
    )

    edges = submission[
        (submission["dataset"] == dataset) &
        (submission["row_type"] == "edge")
    ]

    invalid_source = (
        ~edges["source_id"].isin(nodes)
    )

    invalid_target = (
        ~edges["target_id"].isin(nodes)
    )

    print(
        f"{dataset}: "
        f"invalid source IDs = "
        f"{invalid_source.sum()}, "
        f"invalid target IDs = "
        f"{invalid_target.sum()}"
    )


# ============================================================
# 12. FINAL FILE CHECK
# ============================================================

print("\n========================================")
print("DONE")
print("========================================")

print(
    "submission.csv exists:",
    os.path.exists(submission_path)
)

if os.path.exists(submission_path):
    print(
        "File size:",
        round(
            os.path.getsize(submission_path) / 1024,
            2
        ),
        "KB"
    )

print("\nFirst 10 rows:")
display(
    submission.head(10)
)
```
